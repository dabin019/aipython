```python

coffee_price = 2000
latte_price  = 3000
greentea_price = 2500
juice_price  = 2800

coffee_cnt = int(input("커피 주문수: "))
latte_cnt = int(input("라떼 주문수: "))
greentea_cnt = int(input("녹차 주문수: "))
juice_cnt = int(input("주스 주문수: "))

total = (coffee_price * coffee_cnt +
         latte_price * latte_cnt +
         greentea_price * greentea_cnt +
         juice_price * juice_cnt)

print(f"주문금액: {total}원")
```
